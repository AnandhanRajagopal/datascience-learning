```python
## Anandhan - 2025166043 - DSLab Mock Test - 20/11/2025

import numpy as np
import pandas as pd
import seaborn as sns
import os
from scipy import stats
import matplotlib.pyplot as plt

file_path = "BookDataset.csv"
data = pd.read_csv(file_path)

df = pd.DataFrame(data)
print(df.head())
```

```python
print("Shape of the dataset\n", df.shape)
```

```python
print("First Five Rows \n", df.head(5))
print("Last Five Rows\n", df.tail(5))
```

```python
print("The Most Expensive Car Company Name: ", df.loc[df['price'].idxmax()]["Company"])
```

```python
print(df.isnull().sum())
cleaned_dataset = df.dropna()
cleaned_dataset.to_csv("Cleaned_DataSet.csv", index = False)
#pd.to_csv(cleaned_dataset)
```

```python
df["capacity"] = df["horsepower"]/df["average-mileage"]
print(df["capacity"])
```

```python
print(df[df.Company == "bmw"])
```

```python
print("Total cars per company: ", df['Company'].value_counts())
```

```python
print("Each company highest price cars", df.groupby("Company")["price"].max())
```

```python

print("Average mileage for each car company", df.groupby("Company")["average-mileage"].mean())
```

```python
print("Sorted Cars by Price", df.sort_values(by = "price"))
```

```python
sns.boxplot(x="Company", y="price", data=df)
plt.xticks(rotation=90)
plt.title("Boxplot of Price by Company")
plt.show()

sns.scatterplot(x="average-mileage", y="price", hue="Company", data=df)
plt.title("Mileage vs Price")
plt.show()

sns.kdeplot(df["price"], fill=True)
plt.title("Price Density Plot")
plt.show()

plt.hist(df["price"], bins=20)
plt.title("Histogram of Price")
plt.show()
```

```python
mct = df.groupby("Company")["horsepower"].agg(["mean", "median", lambda x: x.mode().iloc[0]])
print(mct)
print("Mean: ", df["horsepower"].mean())
print("Median: ", df["horsepower"].median())
print("Mode: ", df["horsepower"].mode())
```

```python
print("Covariance", df[["average-mileage", "price"]].cov())
print("Correlation", df[["average-mileage", "price"]].corr())
```

```python
matrix = df[["horsepower", "average-mileage", "price"]].corr()
eigen_values, eigen_vectors = np.linalg.eig(matrix)
print("Eigen Value", eigen_values)
print("Eigen Vectors", eigen_vectors)

```
