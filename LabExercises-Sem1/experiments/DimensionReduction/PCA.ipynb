```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Example dataset
X = np.array([
    [2.5, 2.4],
    [0.5, 0.7],
    [2.2, 2.9],
    [1.9, 2.2],
    [3.1, 3.0],
    [2.3, 2.7],
    [2,   1.6],
    [1,   1.1],
    [1.5, 1.6],
    [1.1, 0.9]
])

pca = PCA(n_components=1)
X_transformed = pca.fit_transform(X)

print("Transformed Matrix", X_transformed)

print("Original shape:", X.shape)
print("Transformed shape:", X_transformed.shape)
print("Principal Component:", pca.components_)
print("Explained Variance:", pca.explained_variance_)
```

```python
import numpy as np

# Step 1: Data
X = np.array([
    [2.5, 2.4],
    [0.5, 0.7],
    [2.2, 2.9],
    [1.9, 2.2],
    [3.1, 3.0],
    [2.3, 2.7],
    [2,   1.6],
    [1,   1.1],
    [1.5, 1.6],
    [1.1, 0.9]
])

# Step 2: Standardize (mean centering)
X_meaned = X - np.mean(X, axis=0)

# Step 3: Covariance matrix
cov_matrix = np.cov(X_meaned, rowvar=False)

# Step 4: Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

# Step 5: Sort by decreasing eigenvalue
sorted_index = np.argsort(eigenvalues)[::-1]
eigenvectors = eigenvectors[:, sorted_index]

# Step 6: Transform data (reduce to 1D)
pc1 = eigenvectors[:, 0]
X_reduced = X_meaned.dot(pc1)

print("PC1 Eigenvector:", pc1)
print("Reduced Values:", X_reduced)
```
