```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA
import numpy as np

X = np.array([
    [2, 3], [3, 4], [4, 5],   # Class 1
    [6, 8], [7, 9], [8, 10]   # Class 2
])
y = np.array([0,0,0, 1,1,1])

lda = LDA(n_components=1)
X_transformed = lda.fit_transform(X, y)

print("Projection Vector (W):", lda.coef_)
print("Transformed Data:", X_transformed)
```

```python
import numpy as np

# class-wise samples
X1 = np.array([[2, 3], [3, 4], [4, 5]])
X2 = np.array([[6, 8], [7, 9], [8, 10]])

# Step 1: Means
m1 = np.mean(X1, axis=0)
m2 = np.mean(X2, axis=0)

# Step 2: Scatter matrices
S1 = (X1 - m1).T @ (X1 - m1)
S2 = (X2 - m2).T @ (X2 - m2) 
Sw = S1 + S2

# Step 3: Use pseudo-inverse (handles singular matrices)
w = np.linalg.pinv(Sw) @ (m2 - m1)

# Step 4: Project the data
projected_X1 = X1 @ w
projected_X2 = X2 @ w

print("W (projection vector):", w)
print("Projected Class 1:", projected_X1)
print("Projected Class 2:", projected_X2)
```
