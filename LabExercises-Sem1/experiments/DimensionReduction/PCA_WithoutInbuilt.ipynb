```python
import numpy as np
X = np.array([[64,71,53,67,55,58,77,57,56,51,76,68], [57,59,49,62,51,50,55,48,42,42,61,57],], dtype=float)

print("Original Data: \n", X)
```

```python
X_meaned = X-np.mean(X, axis=0)
print("\n Mean Centered Data:\n", X_meaned)
```

```python
cov_matrix = np.cov(X_meaned.T)
print("\nCovariance Matrix: \n", cov_matrix)
```

```python
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
print("\n Eigen Values:\n", eigenvalues)
print("\n Eigen Vectors: \n", eigenvectors)
```

```python
k = 2
eigenvectors_k = eigenvectors[:,:k]
print(f"\n Top {k} Eigenvectors:\n", eigenvectors_k)
```

```python
X_reduced = X_meaned @ eigenvectors_k
print(f"\nPCA Reduced Data(k = {k}): \n", X_reduced)
```
