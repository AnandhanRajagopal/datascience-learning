```python
import numpy as np
import pandas as pd
import matplotlib as plt

x = [5, 9, 2, 3, 4, 6, 7, 0, 8, 12, 2, 9]

print(x[1])

print(x[1:4])

print([x[i] for i in [1,2,5]])

print([x[i] for i in range(0,5)] + [x[i] for i in range(9,12)])

print([x[i] for i in range(len(x)) if i not in range(10,12)])
```

```python
#2. Milk Sales - Statistical analysis
y = [33,44,29,16,25,45,33,19,54,22,21,49,11,24,56]

data = np.array(y).reshape(5,3)

df = pd.DataFrame(data, columns=["Shop1", "Shop2","Shop3"], index=["Mon", "Tue","Wed", "Thu", "Fri"])

print(df)

print("\n Statistical Summary for Each Day \n")
print(df.apply(lambda r: r.describe(), axis=0))

print("\n Statistical Summary for Each Shop \n")
print(df.apply(lambda r: r.describe(), axis=1))
```

```python
# 3. Matrix Operations
x = np.array([[3,2], [-1,1]])
y = np.array([[1,4,0], [0,1,-1]])

print(" 2* x \n", 2* x)
print("\n x* x \n", x* x)
print("\n x %*% x \n", x @ x)
print("\n x%*% y \n", x @ y)
print("\n Tanspose \n", y.T)
print("\n Inverser \n", np.linalg.inv(x))

```

```python
# 4. Subscript Operations

x = np.array([[3,2], [-1,1]])
y = np.array([[1,4,0], [0,1,-1]])

print("\n a) x[1,] --> Print First Row \n", x[0,:] )
print("\n b) x[2,] --> Print Second Row \n", x[1,:] )
print("\n c) x[,2] --> Print Second Column \n", x[:,1] )
print("\n d) y[1,2] --> Element at row1 col2 \n", y[0,1] )
print("\n e) y[,2:3] --> Print Column 2 to 3 \n", y[:, 1:3] )
```
