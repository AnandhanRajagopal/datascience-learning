```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [2,4,5,7,6]
```

```python
#Line Chart
plt.plot(x,y, marker ='o')
plt.title("Line Chart")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid(True)
plt.show()
```

```python
#Bar Chart

Categories = ['A', 'B', 'C', 'D']
Values = [10, 24, 36, 30]

plt.bar(Categories, Values)
plt.title("Bar Chart")
plt.xlabel("Category")
plt.ylabel("Values")
plt.show()
```

```python
#Horizontal Bar Chart
plt.barh(Categories, Values)
plt.title("Horizontal Bar Chart")
plt.xlabel("Values")
plt.ylabel("Category")
plt.show()
```

```python
#Scatter Plot

scatterX = [5,7,8,7,2,17,2,9]
scatterY = [99,86,87,88,100,86,103,87]

plt.scatter(x,y)
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()

```

```python
#Histogram
import numpy as np
data = np.random.randn(100)

plt.hist(data, bins = 10)
plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
#Pie chart
sizes = [30,25,20,15,10]
labels = ['A', 'B', 'C', 'D', 'E']

plt.pie(sizes, labels = labels, autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show
```

```python
# Box Plot

data = [np.random.normal(0,1,100),(np.random.normal(5,2,100)), np.random.normal(10,3,100)]

plt.boxplot(data)
plt.title("Box Plot")
plt.xlabel("DataSet")
plt.ylabel("Value")
plt.show()
```

```python
# Area Chart
plt.fill_between(x,y, color='lightblue')
plt.plot(x,y)
plt.title("Area Chart")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
# Multiple Line Chart

y1=[1,3,4,6,8]
y2=[2,4,5,7,9]

plt.plot(x, y1, label = "Line 1")
plt.plot(x, y2, label="Line 2")

plt.title("Multiple Line Chart")
plt.legend()
plt.xlabel("X")
plt.ylabel("Values")
plt.show()
```

```python
# Heat Map

data = np.random.rand(5,5)
print(data)

plt.imshow(data, cmap='viridis')
plt.colorbar()
plt.title("HeatMap")
plt.show()
```
